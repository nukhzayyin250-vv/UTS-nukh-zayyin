```python
volume = [120,185,150,110,135,160,145,125,175,190]
total = sum(volume)
rata_rata = sum(volume)/len(volume)
tertinggi = max(volume)
terendah = min(volume)

print("total:", total)
print("rata-rata:", rata_rata)
print("volume tertinggi:", tertinggi)
print("volume terendah:", terendah)
```

```python
import numpy as x

volume = x.array([120,185,150,110,135,160,145,125,175,190])
print("total:", x.sum(volume))
print("rata-rata:", x.mean(volume))
print("volume tertinggi:", x.max(volume))
print("volume terendah:", x.min(volume))
```

```python
import pandas as pd

# Data yang digunakan pada sel sebelumnya
data = {
    "Jam Pengamatan":["07.00","08.00","09.00","10.00","11.00","12.00","13.00","14.00","15.00","16.00"],
    "Volume Kendaraan":[120,185,150,110,135,160,145,125,175,190]
}

# Membuat file CSV terlebih dahulu agar tidak error FileNotFoundError
pd.DataFrame(data).to_csv("volume_lalu_lintas.csv", index=False, sep=";")

# Membaca kembali file CSV yang baru saja dibuat
df = pd.read_csv("volume_lalu_lintas.csv", sep=";")
print(df.head())
print("jumlah baris:", len(df))
```
